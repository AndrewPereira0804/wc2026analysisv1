```python
from pathlib import Path
import sys

root = Path.cwd().resolve()
if not (root / "src").exists():
    root = root.parent

if not (root / "data").exists():
    root = root.parent

if not (root / "outputs").exists():
    root = root.parent

if str(root) not in sys.path:
    sys.path.append(str(root))
```

```python
from src.load_data import *
from src.cleaning import clean_decisions

df = load_decision_data()
df = clean_decisions(df)

#quick check for blanks

if df.isna().any().any():
    raise ValueError("Missing value(s) in the table")

print("Passed")
```

```python
#scores within ranges
if not df["score"].between(-4,4).all():
    raise ValueError("Found invalid score")
if not df["call_correctness"].between(0,1).all():
    raise ValueError("Found invalid call_correctness")
if not df["confidence"].between(0,1).all():
    raise ValueError("Found invalid confidence")
if not df["standard_fit"].between(-1,1).all():
    raise ValueError("Found invalid standard_fit")
print("Passed")
```

```python
#valid booleans
if not df["VAR_involved"].dtype == 'bool':
    raise ValueError("Found invalid field in VAR_involved")
if not df["favored_team_benefited"].dtype == 'bool':
    raise ValueError("Found invalid field in favored_team_benefited")
print("Passed")
```
